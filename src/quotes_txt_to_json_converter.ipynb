```python
import json
```

```python
# Quote formatting:
#
# Quote text
# 
# - Author Name
# % (break before next quote)
```

```python
def parseQuoteBlock(quote_block: str):
    '''
    :quote_block: A string representation of a quote block

    Returns a {quote: '...', author: '...'} dict for the quote. If the author is missing from the quote block, the author field is an empty string.
    '''
    quote_text = ''
    author_text = ''

    # Search through the block for the author marker '\n\n'
    author_marker = '\n\n'
    author_marker_location = quote_block.find(author_marker)
    
    if author_marker_location == -1:
        # If there is no author marker, the quote text is just the entire quote block
        quote_text = quote_block
        author_text = ''
    else:
        quote_text, author_text = quote_block.split(author_marker)
        
    return {'quote': quote_text, 'author': author_text}

    

test_quote_block = 'Praise is extraneous. The object of praise remains what it was — no better and no worse.\n\n- Marcus Aurelius'
parseQuoteBlock(test_quote_block)
```

```python
def processQuotes(quote_txt_path, json_out_path):
    '''
    Reads quotes from the txt file at quote_txt_path, parses them, and saves a json file with their contents to quote_json_out_path.
    '''

    with open(quote_txt_path) as file:
        content = file.read() # Read file content (produces one big string)
        content = content.split("\n%") # Split content into quote blocks
        content = [block.strip() for block in content] # Remove trailing newlines/whitespaces from quote blocks

    parsed_quotes = [parseQuoteBlock(quote_block) for quote_block in content]

    with open(json_out_path, 'w', encoding='utf8') as out_file:
        json.dump(parsed_quotes, out_file, indent=2, ensure_ascii=False)
        print(f'Quote data written to {json_out_path}')


quote_txt_path = "/home/focus/Desktop/Egne citater/quotes"
json_out_path = '/home/focus/Documents/GitHub/Quote-of-the-day/public/quotes_folder/quotes.json'

processQuotes(quote_txt_path, json_out_path)
```
